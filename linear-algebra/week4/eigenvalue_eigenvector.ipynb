```python
import numpy as np
```

```python
A = np.array([[2, 3], [2, 1]])
e1 = np.array([1, 0])
e2 = np.array([0, 1])
```

```python
# Finding Eigenvalues and Eigenvectors in Python
A_eig = np.linalg.eig(A)

print("\n")

print(f"Matrix A:\n{A} \n\nEigenvalues of matrix A:\n{A_eig[0]}\n\nEigenvectors of matrix A:\n{A_eig[1]}")
```

```python
# first eigenvector
A_eig[1][:, 0]
```

```python
# second eigenvector
A_eig[1][:, 1]
```
